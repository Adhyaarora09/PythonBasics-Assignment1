4Que: write a program that checks if a given word is a python keyword.

```python
import keyword

def is_python_keyword(word):
  """
  Checks if a given word is a Python keyword.

  Args:
    word: The word to check.

  Returns:
    True if the word is a keyword, False otherwise.
  """
  return keyword.iskeyword(word)

# Example usage
word1 = "for"
word2 = "hello"
word3 = "True"

print(f"'{word1}' is a keyword: {is_python_keyword(word1)}")
print(f"'{word2}' is a keyword: {is_python_keyword(word2)}")
print(f"'{word3}' is a keyword: {is_python_keyword(word3)}")
```

6Que:- Write a function to demonstrate the behavior of mutable and immutable arguments.

```python
def modify_arguments(immutable_arg, mutable_arg):
    """
    Demonstrates the behavior of mutable and immutable arguments in Python.

    Args:
        immutable_arg: An immutable object (e.g., integer, string, tuple).
        mutable_arg: A mutable object (e.g., list, dictionary).
    """
    print(f"Original immutable argument: {immutable_arg}")
    print(f"Original mutable argument: {mutable_arg}")

    # Attempt to modify the immutable argument
    immutable_arg = immutable_arg + 1  # Reassignment creates a new object
    print(f"Modified immutable argument inside function: {immutable_arg}")

    # Modify the mutable argument
    mutable_arg.append(4)  # Modifies the original object
    print(f"Modified mutable argument inside function: {mutable_arg}")

# Example usage:
immutable_value = 10
mutable_list = [1, 2, 3]

modify_arguments(immutable_value, mutable_list)

print(f"Immutable argument outside function: {immutable_value}")
print(f"Mutable argument outside function: {mutable_list}")
```

7Que:- Write a program that performs basic arithmetic operations on two user-input numbers.

```python
def add(x, y):
  return x + y

def subtract(x, y):
  return x - y

def multiply(x, y):
  return x * y

def divide(x, y):
  if y == 0:
    return "Cannot divide by zero"
  return x / y

def integer_divide(x, y):
  if y == 0:
    return "Cannot divide by zero"
  return x // y

def modulo(x, y):
    if y == 0:
        return "Cannot calculate modulo with zero"
    return x % y

def exponent(x, y):
    return x ** y

num1 = 10
num2 = 5

print("Addition:", add(num1, num2))
print("Subtraction:", subtract(num1, num2))
print("Multiplication:", multiply(num1, num2))
print("Division:", divide(num1, num2))
print("Integer Division:", integer_divide(num1, num2))
print("Modulo:", modulo(num1, num2))
print("Exponent:", exponent(num1, num2))
```

9Que:- Write a Python program to convert user input from string to integer, float, and boolean types.

```python
def convert_input_type(user_input):
    """
    Converts user input to integer, float, or boolean type if possible
  Returns the converted value or the original string if conversion is not possible
    """
    if user_input.isdigit():
        return int(user_input)
    try:
        return float(user_input)
    except ValueError:
        if user_input.lower() == "true":
            return True
        elif user_input.lower() == "false":
            return False
        else:
            return user_input

# Get input from the user
user_input = input("Enter a value: ")

# Convert the input to the appropriate type
converted_value = convert_input_type(user_input)

# Print the converted value and its type
print("Converted value:", converted_value)
print("Type:", type(converted_value))
```

10 Que:- Write code to demonstrate type casting with list elements.

```python
my_list = ["1", "2", "3", "4.5", "hello"]

# Convert string elements to integers
int_list = []
for item in my_list:
    try:
        int_list.append(int(item))
    except ValueError:
        print(f"Cannot convert '{item}' to integer.")

print("List of integers:", int_list)

# Convert string or integer elements to floats
float_list = []
for item in my_list:
    try:
        float_list.append(float(item))
    except ValueError:
        print(f"Cannot convert '{item}' to float.")

print("List of floats:", float_list)

# Convert all elements to strings
str_list = []
for item in my_list:
    str_list.append(str(item))

print("List of strings:", str_list)
```

14Que:- Write a program to reverse a string using a while loop.

```python
def reverse_string(s):
    """Reverses a string using a while loop.

    Args:
        s: The string to reverse.

    Returns:
        The reversed string.
    """
    left = 0
    right = len(s) - 1
    s_list = list(s)  # Convert string to list for in-place modification

    while left < right:
        s_list[left], s_list[right] = s_list[right], s_list[left]
        left += 1
        right -= 1

    return "".join(s_list)  # Convert list back to string


# Example usage
string = "hello"
reversed_string = reverse_string(string)
print(f"Original string: {string}")
print(f"Reversed string: {reversed_string}")
```

15Que:- Write a Python program to calculate the factorial of a number provided by the user using a while loop.

```python
number = int(input("Enter a non-negative integer: "))

if number < 0:
    print("Factorial is not defined for negative numbers.")
elif number == 0:
    print("The factorial of 0 is 1")
else:
    factorial = 1
    while number > 0:
        factorial *= number
        number -= 1
    print("The factorial is", factorial)
```
